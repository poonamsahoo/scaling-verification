```python
import datasets
import utils
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import wandb
import nb
```

```python
hub_name = "amyguan/math500-llama8b-10-10-80-nb"
dev_ds = datasets.load_dataset(f"{hub_name}-dev")['data']
val_ds = datasets.load_dataset(f"{hub_name}-val")['data']
test_ds = datasets.load_dataset(f"{hub_name}-test")['data']
```

```python
# dev_ds['ratio_correct'] = 

_, verifier_names = utils.extract_scores_matrix(dev_ds)

print("\nExtracting features from datasets...")
X_dev, y_dev_difficulty = nb.extract_features_from_dataset(dev_ds, verifier_names)
X_val, y_val_difficulty = nb.extract_features_from_dataset(val_ds, verifier_names)
X_test, y_test_difficulty = nb.extract_features_from_dataset(test_ds, verifier_names)
```

```python
# dev_ds['nb_diff'], dev_ds['level']
```

```python
# Create DataFrames for dev, val, and test sets
df_dev = pd.DataFrame()
df_dev['nb_diff'] = dev_ds['nb_diff']
df_dev['level'] = dev_ds['level']
df_dev['oracle_diff'] = y_dev_difficulty

df_val = pd.DataFrame()
df_val['nb_diff'] = val_ds['nb_diff']
df_val['level'] = val_ds['level']
df_val['oracle_diff'] = y_val_difficulty

df_test = pd.DataFrame()
df_test['nb_diff'] = test_ds['nb_diff']
df_test['level'] = test_ds['level']
df_test['oracle_diff'] = y_test_difficulty
```

```python
def print_correlations(df):
    print('Corr of level and nb_diff: ', df['level'].corr(df['nb_diff']))
    print('Corr of oracle_diff and nb_diff: ', df['oracle_diff'].corr(df['nb_diff']))
    print('Corr of level and oracle_diff: ', df['level'].corr(df['oracle_diff']))

print('\nDev')
print_correlations(df_dev)
print('\nVal')
print_correlations(df_val)
print('\nTest')
print_correlations(df_test)
```
